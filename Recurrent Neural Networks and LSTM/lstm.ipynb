# **LONG SHORT-TERM MEMORY**

## **OS PROBLEMAS DO GRADIENTE**

Dentro desses materiais, descrevi diversas formas de tentar segurar a explosão ou o desaparecimento do gradiente. Todas essas técnicas são aplicáveis tanto às redes *feedforward* quanto às *RNNs*, mas as *RNNs* possuem algumas propriedades únicas e potenciais técnicas de mitigação. 

Como comentado no material anterior, as *RNNs* têm a propriedade de que a *BPTT* resulta na multiplicação do erro pelo mesmo peso repetidamente, devido ao compartilhamento de pesos entre os passos de tempo. Com um número suficiente de passos, essa multiplicação repetitiva pode levar ao desaparecimento ou a explosão do gradiente. A única maneira de evitar o desaparecimento e a explosão de gradientes é usar pesos com valor 1, o que meio que anula o propósito, pois queremos poder ajustar os pesos do modelo. No entanto, essa ideia pode ser usada para criar uma unidade recorrente mais complexa, que utiliza uma técnica chamada **carrossel de erro constante (*CEC*)**. O uso do *CEC* resulta em um comportamento similar a ter pesos de valor 1 durante a *backpropagation*.
